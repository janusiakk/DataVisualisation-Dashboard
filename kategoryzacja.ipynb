```python
import json
import pandas as pd

# Wczytanie danych z pliku
with open('Dane\PC-SUM-K-03.01.2026.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

df = pd.DataFrame(data)

def classify_activity(row):
    app = str(row.get('app', '')).lower()
    title = str(row.get('title', '')).lower()
    combined = app + " " + title
    
    # 1. Studia (Data Science & Engineering focus)
    # Dodano MATLAB i LeON na podstawie podglądu pliku
    ds_keywords = [
        'python', 'pandas', 'numpy', 'scikit', 'sklearn', 'jupyter', 'colab', 
        'code.exe', 'visual studio', 'vscode', 'pycharm', 'rstudio', 
        'matlab', 'mathworks', 'simulink', 'numeryczne', 'metody numeryczne',
        'sql', 'kaggle', 'github', 'git', 'stack overflow', 'stackoverflow', 
        'data science', 'machine learning', 'ai', 'sztuczna inteligencja',
        'gemini', 'chatgpt', 'openai', 'claude', 'leon', 'usos', 'politechnika',
        'pochodna', 'całka', 'zapiski', 'laboratorium', 'projekt', 'studia',
        '.py', '.ipynb', '.m', '.cpp', '.c', 'json crack'
    ]
    if any(k in combined for k in ds_keywords):
        return 'studia'

    # 2. Komunikatory
    comm_keywords = [
        'discord', 'messenger', 'whatsapp', 'signal', 'telegram', 'teams', 
        'zoom', 'slack', 'skype', 'outlook', 'poczta', 'mail', 'thunderbird',
        'facebook', 'instagram', 'twitter', 'x.com', 'linkedin'
    ]
    if any(k in combined for k in comm_keywords):
        return 'Komunikatory'

    # 3. Gry komputerowe
    game_keywords = [
        'steam', 'epic games', 'riot', 'league of legends', 'lol', 'valorant', 
        'cs', 'counter-strike', 'minecraft', 'roblox', 'fortnite', 'gta', 
        'wiedźmin', 'cyberpunk', 'blizzard', 'battlenet', 'ubisoft', 'game', 
        'gra', 'binding of isaac', 'hollow knight', 'clash of clans'
    ]
    if any(k in combined for k in game_keywords):
        return 'gry komputerowe'

    # 4. Oglądanie filmów
    video_keywords = [
        'youtube', 'youtu.be', 'netflix', 'hbo', 'disney', 'prime video', 
        'twitch', 'vlc', 'potplayer', 'cda', 'film', 'movie', 'serial', 
        'wmplayer', 'windows media player'
    ]
    if any(k in combined for k in video_keywords):
        return 'oglądanie filmów'

    # 5. Muzyka
    music_keywords = [
        'spotify', 'tidal', 'apple music', 'soundcloud', 'deezer', 'foobar', 
        'winamp', 'music', 'muzyka'
    ]
    if any(k in combined for k in music_keywords):
        return 'muzyka'

    # 6. Nauka (Ogólna - jeśli nie wpadło do Data Science)
    learn_keywords = [
        'wikipedia', 'wiki', 'kurs', 'course', 'udemy', 'coursera', 'edx', 
        'duolingo', 'tłumacz', 'translate', 'słownik', 'dictionary', 'notion', 'obsidian'
    ]
    if any(k in combined for k in learn_keywords):
        return 'nauka'

    # 7. Inne (System, Przeglądarka ogólna, Zakupy, Zdjęcia itp.)
    return 'inne'

# Aplikowanie klasyfikacji
df['category'] = df.apply(classify_activity, axis=1)

# Zapis do pliku JSON
output_path = 'Dane\PC-SUM-K-03.-01.2026.json'
df.to_json(output_path, orient='records', force_ascii=False, indent=4)

print(f"Dane zostały sklasyfikowane i zapisane w: {output_path}")
```

```python
import os
import pandas as pd

# --- KONFIGURACJA ---
# Zmień te ścieżki na swoje
PLIK_WEJSCIOWY = 'Dane/PC-SUM-K-03.01.2026.json' 
FOLDER_WYJSCIOWY = 'Dane Dash'
PLIK_WYJSCIOWY = os.path.join(FOLDER_WYJSCIOWY, 'Dane1.json')

# TWOJE STAŁE WARTOŚCI
NAZWA_OSOBY = "A"
NAZWA_URZADZENIA = "PC"

def przetworz_dane_aw():
    # 1. Upewnienie się, że folder docelowy istnieje
    if not os.path.exists(FOLDER_WYJSCIOWY):
        os.makedirs(FOLDER_WYJSCIOWY)

    try:
        # 2. Odczyt JSON
        df = pd.read_json(PLIK_WEJSCIOWY)
        
        # 3. Dodanie kolumn 'osoba' oraz 'device'
        # Pandas automatycznie wypełni wszystkie wiersze tą samą wartością
        df['osoba'] = NAZWA_OSOBY
        df['device'] = NAZWA_URZADZENIA

        # 4. Zapis do nowego pliku
        # orient='records' i indent=4 tworzą czytelny, standardowy plik JSON
        df.to_json(PLIK_WYJSCIOWY, orient='records', indent=4, force_ascii=False)
        
        print(f"Sukces! Plik zapisany w: {PLIK_WYJSCIOWY}")
        print("\nPodgląd danych:")
        print(df[['osoba', 'device']].head(3)) # Pokazuje tylko nowe kolumny dla testu

    except Exception as e:
        print(f"Błąd podczas przetwarzania: {e}")

if __name__ == "__main__":
    przetworz_dane_aw()
```
